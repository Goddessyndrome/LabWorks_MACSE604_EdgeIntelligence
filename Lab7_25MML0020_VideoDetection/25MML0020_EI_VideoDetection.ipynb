Importing Libraries

```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Starting live video and checking for security

```python
# Opening webcam (0 = default camera)
cap = cv2.VideoCapture(0)

# Security check to confirm camera access
if not cap.isOpened():
    print("Camera access failed")
else:
    print("Camera started successfully")
```

Displaying video properties

```python
# Getting video properties
fps = cap.get(cv2.CAP_PROP_FPS)
width = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

print("FPS:", fps)
print("Width:", width)
print("Height:", height)
```

Counting frames while capturing 

```python
frame_count = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_count += 1
    cv2.imshow("Live Video", frame)

    if cv2.waitKey(1) == ord('q'):
        break

print("Total frames captured:", frame_count)

cap.release()
cv2.destroyAllWindows()
```

Capturing one frame and disaplying it 

```python
cap = cv2.VideoCapture(0)

ret, frame = cap.read()

print("Frame shape:", frame.shape)

plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
plt.title("Captured Frame")
plt.axis("off")

cap.release()
```

Converting to grayscale and printing matrix

```python
# Converting frame to grayscale
gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Frame")
plt.axis("off")

# Print grayscale matrix
print("Grayscale matrix:")
print(gray)
```

Printing pixel values of both colored and gray scales

```python
# Selecting a pixel location
x = 100
y = 100

# Colored pixel value
print("RGB pixel value:", frame[y, x])

# Grayscale pixel value
print("Gray pixel value:", gray[y, x])
```

Saving every 100th frame 

```python
cap = cv2.VideoCapture(0)

count = 0
saved = 0

while True:
    ret, frame = cap.read()
    if not ret:
        break

    if count % 100 == 0:
        filename = f"frame_{saved}.jpg"
        cv2.imwrite(filename, frame)
        saved += 1

    count += 1

    cv2.imshow("Video", frame)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

Showing frame dimensions 

```python
print("Frame dimensions:", frame.shape)
```

Drawing various shapes using CV2

```python
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Rectangle
    cv2.rectangle(frame, (100,100), (300,300), (0,255,0), 3)

    # Circle
    cv2.circle(frame, (200,200), 50, (255,0,0), 3)

    # Text
    cv2.putText(frame, "OpenCV", (50,50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2)

    cv2.imshow("Shapes", frame)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

Edge Detection using Canny 

```python
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    edges = cv2.Canny(frame, 100, 200)

    cv2.imshow("Edges", edges)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

Negative of video 

```python
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Unable to access camera")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Increasing contrast for sharper appearance
    enhanced = cv2.convertScaleAbs(frame, alpha=1.5, beta=0)

    # Creating negative image
    negative = 255 - enhanced

    cv2.imshow("Sharp Negative Video", negative)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
